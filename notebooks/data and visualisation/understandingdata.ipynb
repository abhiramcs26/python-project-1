```python
month=['JAN' ,'FEB' ,'MAR', 'APR' , 'MAY' , 'JUN' , 'JUL',  'AUG'  , 'SEP' , 'OCT' , 'NOV' , 'DEC' ]
invyear=~df['YEAR'].between(1901,2017)
invrainfall=(df[month].isnull().any(axis=1) | (df[month]<0).any(axis=1))
invrec=invyear | invrainfall

```

```python
meanrainfall =df['ANNUAL'].mean()
anomaly=df['ANNUAL']-meanrainfall
rollingmean=anomaly.rolling(window=10).mean()
std=anomaly.std()
plt.figure(figsize=(12,6))
plt.bar(df['YEAR'],anomaly,color=np.where(anomaly>=0,'blue','red'),width=0.9)
plt.plot(df['YEAR'],rollingmean,color='black',linewidth=2 ,label='10-years rolling mean')
plt.axhline(0,color='grey',linewidth=1)
plt.axhline(1.5*std,color='green',linestyle='--',label='+1.5 standard deviation ')
plt.axhline(-1.5*std,color='orange',linestyle='--',label='=1.5 standard deviation ')
plt.title('Annual Rainfall Anomaly in Kerala (1901-2017)')
plt.xlabel('Year')
plt.ylabel('Rainfall Anomaly in mm')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
decaderainfall=df.assign(decade=(df['YEAR']//10)*10).groupby('decade')[month].mean()
decaderainfall
```

```python
decaderainfall=df.assign(decade=(df['YEAR']//10)*10).groupby('decade')[month].mean()
plt.figure(figsize=(16,8))
sns.heatmap(decaderainfall,cmap='YlGnBu',annot=True,fmt='.0f',linewidth=0.5)
plt.title('Decade wise Monthly Rainfall in Kerala 1901-2017')
plt.xlabel('Month')
plt.ylabel('Decade')
plt.tight_layout()
plt.show()
```
